```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
```

```python
make_classification(n_samples=1000, n_features=10, n_redundant=5, n_informative=5, n_classes=2, random_state=1)
```

```python
x, y = make_classification(n_samples=1000, n_features=10, n_redundant=5, n_informative=5, n_classes=2, random_state=1)
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3, random_state=1)
```

```python
model = LogisticRegression()
```

```python
model
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
# Evaluation matrics 
    
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print(confusion_matrix(y_test, y_pred)) # confusion matrics 
```

```python
print(accuracy_score(y_test, y_pred)) # accuracy score for tyhe model
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# get pradicted probablities for class 1
y_pred_proba = model.predict_proba(x_test)[:, 1] # probablity for one class 
```

```python
y_pred_proba
```

```python
fpr, tpr, threshilds = roc_curve(y_test, y_pred_proba)
```

```python
# calculate auc score 
roc_aoc = auc(fpr,tpr)
roc_aoc
```

```python
# plot ro-aoc curve

plt.Figure(figsize=(8,6)) # create a new figure with a specified size
plt.plot(fpr, tpr, color = 'darkorange', linewidth =2, label = 'ROC curve (area = %0.2f)'% roc_aoc) #plot the roc and aoc
plt.plot([0,1],[0,1], color = 'navy', linewidth =2, linestyle = '--') # plot the diogonal line representing random gasa
plt.xlim([0.0,1.0]) # set x-axis limits
plt.ylim([0.0,1.05]) # set y axis limits
plt.xlabel('false posative rate') # set x-axis label
plt.ylabel('true posative rate') # set y-axis label
plt.title('Receiver operating Characteristic (ROC) Curve') # set plot titel
plt.legend(loc='lower right') # Add Legend to the plot
plt.show() # for show the plot
```

```python
# precsion recall accuracy tradoff to decide optimal thrsehold
# by default sklearn 0.5 use thrsehold
```

```python
# import impotent libraries
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.metrics import precision_score, recall_score, accuracy_score

# calculate precision, recall, accuracy for different threshold probabilitics 
thresholds = np.linspace(0,1,100)
precisions = []
recalls = []
accuracies = []

for threshold in thresholds:
    y_pred_thresholds = (y_pred_proba >= threshold).astype(int)
    precision = precision_score(y_test, y_pred_thresholds)
    recall = recall_score(y_test, y_pred_thresholds)
    accuracy = accuracy_score(y_test, y_pred_thresholds)
    precisions.append(precision)
    recalls.append(recall)
    accuracies.append(accuracy)

# x or y are same so i am just chaking for same x==y 
print(len(thresholds))
print(len(precisions))
print(len(recalls))
print(len(accuracies))

#plot precision, recall, and accuracy against threshold probabilitics 
plt.Figure(figsize=(10, 6))
plt.plot(thresholds, precisions, label ='precision')
plt.plot(thresholds, recalls, label ='recall')
plt.plot(thresholds, accuracies, label = 'accuracy')
plt.xlabel(' Threshold Probability')
plt.ylabel('Score')
plt.title('Precision, Recall, and Accuracy vs. Threshold probability')
plt.legend()
plt.grid(True)
plt.show()
```
