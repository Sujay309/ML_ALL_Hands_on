```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
from sklearn.datasets import fetch_california_housing
data = fetch_california_housing()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['price'] = data.target
```

```python
df
```

```python
df.shape
```

```python
# only 20 persent of the data becose the vesulization perpas 
df = df.sample(frac=0.20)
```

```python
df
```

```python
# independent and dependent features 

x =df.iloc[:,:-1]
y= df.iloc[:,-1]
```

```python
x
```

```python
y
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.20, random_state=1)
x_train.shape, x_test.shape
```

```python
# decesion tree regresser 
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
model
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
r2_score(y_pred, y_test)
```

```python
# hyper parameter teuning

params = {'criterion':['squared_error', 'friedman_mse', 'absolute_error', 'poisson'],
          'splitter' :['best', 'random'],
          'max_depth':[1,2,3,4,5,6],
          'max_features' :['auto', 'sqrt', 'auto']}

regressor = DecisionTreeRegressor()
```

```python
from sklearn.model_selection import GridSearchCV
model = GridSearchCV(regressor, param_grid=params, cv= 3, scoring='neg_mean_squared_error')
```

```python
model.fit(x_train, y_train)
```

```python
model.best_params_
```

```python
final_model = DecisionTreeRegressor(criterion = 'poisson',max_depth= 6,max_features = 'sqrt',splitter= 'best')
```

```python
final_model
```

```python
final_model.fit(x_train, y_train)
```

```python
# visuailised the decision tree 
from sklearn import tree
plt.Figure(figsize=(12, 10))
tree.plot_tree(final_model, filled =True)
```

```python
y_pred = final_model.predict(x_test)
```

```python
r2_score(y_pred, y_test)
```
