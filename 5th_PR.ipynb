```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#generatic synthetic data points
np.random.seed(1)
```

```python
# generating synthetic data pionts
x = 2*np.random.rand(100,1)
```

```python
x
```

```python
y = 4+3*x+1.5*x**2+np.random.randn(100,1)
```

```python
y
```

```python
#train test split
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, Y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
x_train.shape, y_train.shape
```

```python
# perfrom polynomial regression 
from sklearn.preprocessing import PolynomialFeatures
```

```python
degree = 2
poly_features = PolynomialFeatures(degree=2, include_bias=False)
x_poly_train = poly_features.fit_transform(x_train)
```

```python
x_poly_train
```

```python
#implement linear regrasion 
from sklearn.linear_model import LinearRegression
```

```python
ploy_reg = LinearRegression()
```

```python
ploy_reg
```

```python
ploy_reg.fit(x_poly_train, y_train)
```

```python
ploy_reg.coef_
```

```python
ploy_reg.intercept_
```

```python
#making pradiction on train data
y_poly_pred = ploy_reg.predict(x_poly_train)
```

```python
#evaluation matrix 
from sklearn.metrics import mean_squared_error
```

```python
mse_train = mean_squared_error(y_train, y_poly_pred)
print(f'mean squared error on training data : {mse_train}')
```

```python
plt.scatter(x_train, y_train, label = "training")
```

```python
plt.scatter(x_train, y_train, label = "training")
x_range = np.linspace(0, 2, 100).reshape(-1, 1)
x_range_poly = poly_features.transform(x_range)
plt.plot(x_range, ploy_reg.predict(x_range_poly), color = 'r', label=f'polynomial Regression')
plt.xlabel('x')
plt.ylabel('y')
plt.title("polynomial Regression")
plt.legend()
plt.show()
```
