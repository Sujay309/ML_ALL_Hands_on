```python
# import all nasasary lybaris 

import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

```python
df
```

```python
df.tail()
```

```python
df.target.unique()
```

```python
# for binary class classification 

dff = df[df['target'] != 2]
```

```python
dff
```

```python
dff.target.unique()
```

```python
# split to x and y 

x = dff.iloc[:, :-1]
y = dff.iloc[:, -1]
```

```python
x
```

```python
y
```

```python
# train test split 
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test =train_test_split(x,y, test_size=0.2, random_state=1)
```

```python
x_test.shape, x_train.shape
```

```python
# model traning

from sklearn.linear_model import LogisticRegression 
classifier = LogisticRegression()
```

```python
classifier
```

```python
classifier.fit(x_train,y_train)
```

```python
y_pred = classifier.predict(x_test)
```

```python
y_pred
```

```python
#correspding to higher probability class is predicted, cutoff 0.5.
classifier.predict_proba(x_test)
```

```python
# Evaluation matrics 
    
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
print(confusion_matrix(y_test, y_pred)) # confusion matrics 
```

```python
print(accuracy_score(y_test, y_pred)) # accuracy score for tyhe model
```

```python
print(classification_report(y_test, y_pred))

# first tow rows, 0 and 1 >> precision, recoll, f1 score and on of data point of respective class
# macro avg>> for each class take simple avg>> avg precison of class 0 and class 1 will be macro avg
# similarly for recall, f1 scor, support
# weighted avg is used in case of class importance to performance on dominant class.
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# get pradicted probablities for class 1
classifier.predict_proba(x_test)[:, 1] # probablity for one class 
```

```python
y_pred_proba = classifier.predict_proba(x_test)[:, 1]
```

```python
# reciver oprating characteristics curve
# input : y_test, y_pred_proba. it will return TPR, FPR with different cutoff of probabilities
# how well your model is abel distinguish between tow clases
fpr, tpr, threshilds = roc_curve(y_test, y_pred_proba)
```

```python
fpr
```

```python
tpr
```

```python
threshilds
```

```python
# calculate auc score 
roc_aoc = auc(fpr,tpr)
roc_aoc
```

```python
# plot ro-aoc curve

plt.Figure(figsize=(8,6)) # create a new figure with a specified size
plt.plot(fpr, tpr, color = 'darkorange', linewidth =2, label = 'ROC curve (area = %0.2f)'% roc_aoc) #plot the roc and aoc
plt.plot([0,1],[0,1], color = 'navy', linewidth =2, linestyle = '--') # plot the diogonal line representing random gasa
plt.xlim([0.0,1.0]) # set x-axis limits
plt.ylim([0.0,1.05]) # set y axis limits
plt.xlabel('false posative rate') # set x-axis label
plt.ylabel('true posative rate') # set y-axis label
plt.title('Receiver operating Characteristic (ROC) Curve') # set plot titel
plt.legend(loc='lower right') # Add Legend to the plot
plt.show() # for show the plot
```
