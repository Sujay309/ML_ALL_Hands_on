```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

import warnings 
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('mpg')
```

```python
df.head()
```

```python
df.drop("name", axis=1, inplace=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['horsepower'].median()
```

```python
df['horsepower']=df['horsepower'].fillna(df["horsepower"].median())
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.dtypes
```

```python
df['origin'].value_counts()
```

```python
df['origin'].map({"usa":1, "japan":2, "europe":3})
```

```python
df['origin'] = df['origin'].map({"usa":1, "japan":2, "europe":3})
```

```python
df['origin'] = df['origin'].astype(int)
```

```python
df.dtypes
```

```python
# separate X and y

x = df.drop('mpg', axis=1)
y = df['mpg']
```

```python
x
```

```python
y
```

```python
# train test splite 

from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=1)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
# simple linear regration  model

from sklearn.linear_model import LinearRegression
```

```python
r_model = LinearRegression()

r_model
```

```python
r_model.fit(x_train, y_train)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(i, col_name)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f"the cofficient for {col_name} is {r_model.coef_[i]}")
```

```python
# coefficients are relatively smaller, if independent variable changes slightly thare will be not much difference in prediction.
# this is sometime is called smoother model 
```

```python
from sklearn.metrics import r2_score
```

```python
y_pred_lr = r_model.predict(x_test)
```

```python
r2_linear = r2_score(y_test, y_pred_lr)
print(f"r square of linear regression {r2_linear}")
```

```python
# redge regression 

from sklearn.linear_model import Ridge
```

```python
r_r_model = Ridge(alpha=0.1)
```

```python
r_r_model.fit(x_train, y_train)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f"the cofficient for {col_name} is {r_r_model.coef_[i]}")
```

```python
y_pred_ridge = r_r_model.predict(x_test)
r2_ridge = r2_score(y_test, y_pred_ridge)
print(f"r-square score for ridge regression: {r2_ridge}")
```

```python
#we dont see much variation in coeff of ridge regression as compared to linear regression
```

```python
#lasso redgession
```

```python
from sklearn.linear_model import Lasso
```

```python
l_r_model = Lasso(alpha=0.5)
l_r_model.fit(x_train, y_train)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f"the cofficient for {col_name} is {l_r_model.coef_[i]}")
```

```python
# 3 features coefficients are 0, lasso helps in feature selaction 
```

```python
y_pred_lasso = l_r_model.predict(x_test)
r2_lasso = r2_score(y_test, y_pred_lasso)
print(f"r-square score for lasso regression: {r2_lasso}")
```

```python
# elastik regresion 

from sklearn.linear_model import ElasticNet
```

```python
e_r_model = ElasticNet(alpha=1, l1_ratio=0.5)
e_r_model.fit(x_train, y_train)
```

```python
for i, col_name in enumerate(x_train.columns):
    print(f"the cofficient for {col_name} is {e_r_model.coef_[i]}")
```

```python
y_pred_elastic = e_r_model.predict(x_test)
r2_elastic = r2_score(y_test, y_pred_elastic)
print(f"r-square score for elastic net regression: {r2_elastic}")
```

```python
#cross validation lasso
```

```python
from sklearn.linear_model import LassoCV
Lasso_cv=LassoCV(cv=5)
Lasso_cv.fit(x_train,y_train)

y_pred_lcv=Lasso_cv.predict(x_test)


score = r2_score(y_test, y_pred_lcv)
print("R2 score", score)
```

```python
Lasso_cv.get_params()
```

```python
##cross validation ridge

from sklearn.linear_model import RidgeCV
ridge_cv=RidgeCV(cv=5)
ridge_cv.fit(x_train,y_train)

y_pred_rcv=ridge_cv.predict(x_test)


score = r2_score(y_test, y_pred_rcv)
print("R2 score", score)
```

```python
ridge_cv.get_params()
```

```python
# elastiknet cross validation 

from sklearn.linear_model import ElasticNetCV
elastic_cv = ElasticNetCV(cv=5)
elastic_cv.fit(x_train, y_train)

y_pred_ecv = elastic_cv.predict(x_test)

score = r2_score(y_test, y_pred_ecv)
print("R2 score", score)
```

```python
elastic_cv.get_params()
```
